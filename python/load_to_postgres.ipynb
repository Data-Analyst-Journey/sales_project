```python
#importing data 

import psycopg2
import pandas as pd
from sqlalchemy import create_engine

#----------------------------------------------
# 1. Connection to PostgreSQL (without URL)
#----------------------------------------------
DB_NAME = "sales_db"
DB_USER = "postgres"
DB_PASSWORD = "@Taha1391"
DB_HOST = "localhost"
DB_PORT = "5432"

# Connect directly using psycopg2
conn = psycopg2.connect(
    host = DB_HOST,
    database=DB_NAME,
    user=DB_USER,
    password=DB_PASSWORD,
    port=DB_PORT
)
print("✅ Connected to PostgreSQL")

#Create engine from connection
engine = create_engine(f"postgresql://{DB_USER}:{DB_PASSWORD.replace('@', '%40')}@{DB_HOST}:{DB_PORT}/{DB_NAME}")

#---------------------------------------------------
# 2. Load cleaned data
#---------------------------------------------------
customers = pd.read_csv('customers_cleaned.csv')
products = pd.read_csv('products_cleaned.csv')
sales = pd.read_csv('sales_cleaned.csv')

print(f"✅ Customers: {len(customers)} rows")
print(f"✅ Products:  {len(products)} rows")
print(f"✅ Sales:     {len(sales)} rows")

#--------------------------------------------------
# 3. Write to PostgreSQL
#--------------------------------------------------
customers.to_sql('customers', engine, if_exists='replace', index = False)
products.to_sql('products', engine, if_exists='replace', index = False)
sales.to_sql('sales', engine, if_exists='replace', index = False)

print("✅ Data loaded to PostgreSQL successfully!")
```
