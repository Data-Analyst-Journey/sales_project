# Sales Data Cleaning Pipeline
This notebook cleans and prepares sales data for analysis and modeling.
- Input: 'customers.csv' , 'products.csv' , 'sales.csv'   
- Output: 'customers_cleaned.csv' , 'products_cleaned.csv' , 'sales_cleaned.csv'
#Data Cleaning and preparation for sales analytics project


```python
import pandas as pd
import numpy as np
from datetime import datetime
```

```python
#---------------------------------------------------------------
# 1.Load raw data
#---------------------------------------------------------------
print("=" * 60)
print("SALES DATA CLEANING PIPELINE")
print(f"Start time: {datetime.now()}")
print("=" * 60)

# Load customer,product, and sales data from CSV files
customers = pd.read_csv('customers.csv')
products = pd.read_csv('products.csv')
sales = pd.read_csv('sales.csv')

print(f"\n Initial data shapes:")
print(f"  - Customers: {len(customers)} rows")
print(f"  - Products: {len(products)}rows")
print(f"  - Sales: {len(sales)}rows")
```

```python
# ----------------------------------------------
# 2.Clean customers
# ----------------------------------------------
print("\n Cleaning customers ...")

# Track removed rows
initial_customers = len(customers)

#Standardize city names
customers['city'] = customers['city'].str.strip().str.title()

#Remove invalid ages (using IQR method)
Q1 = customers['age'].quantile(0.25)
Q3 = customers['age'].quantile(0.75)

IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
customers = customers[(customers['age'] >= lower_bound) & (customers['age'] <= upper_bound)]

#Standardize gender
customers['gender'] = customers['gender'].str.lower().str.strip()
customers = customers[customers['gender'].isin(['men' , 'women'])]

#Remove duplicates
customers = customers.drop_duplicates(subset = ['customer_id'])


print(f" - Removed {initial_customers - len(customers)} rows")
print(f" Final: {len(customers)} customers")

```

```python
#-------------------------------------------------------------
# 3. Clean products
#-------------------------------------------------------------
print("\n Cleaning products...")

initial_products = len(products)

# Remove invalid prices
products = products[products['base_price']>0 ]

#Standardize categories
products['category'] = products['category'].str.strip().str.title()
products['product_name'] = products['product_name'].str.title()

# Remove duplicates
products = products.drop_duplicates(subset = ['product_id'])

print(f" - Removed {initial_products - len(products)} rows")
print(f" - Final: {len(products)} products")
```

```python
#-----------------------------------------------------
# 4.Clean sales
# -----------------------------------------------------
print("\n Cleaning sales...")

initial_sales = len(sales)

# Convert date to datetime
sales['sale_data'] = pd.to_datetime(sales['sale_date'])

# Remove invalid sales
sales = sales[sales['unit_price'] > 0]
sales = sales[sales['quantity'] > 0]

# Remove outliers in unit_price (IQR method)
Q1= sales['unit_price'].quantile(0.25)
Q3 = sales['unit_price'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
sales = sales[(sales['unit_price'] >=lower_bound) & (sales['unit_price'] <= upper_bound)]

#Standardize platform names
sales['platform'] = sales['platform'].str.strip().str.title()

#Remove duplicates
sales = sales.drop_duplicates(subset = ['sale_id'])

print(f" - Removed {initial_sales - len(sales)} rows")
print(f" - Final: {len(sales)} sales records")
```

```python
# ------------------------------------------------------------------
# 5. Save cleaned data
#-------------------------------------------------------------------
print("\n Saving cleaned data...")

# Export cleaned datasets to new CSV files
customers.to_csv('customers_cleaned.csv',index=False)
products.to_csv('products_cleaned.csv',index=False)
sales.to_csv('sales_cleaned.csv',index=False)

#---------------------------------------------------------------
# 6. Summary report
#---------------------------------------------------------------
print("\n" + "=" * 60)
print(" DATA CLEANING COMPLETE")
print("=" * 60)
print(f"\n Final summary:")
print(f" - customers: {len(customers)}")
print(f" - Products: {len(products)}")
print(f" - Sales: {len(sales)}")
print(f"\n Files saved:")
print(" - customers_cleaned.csv")
print(" - products_cleaned.csv")
print(" - sales_cleaned.csv")
print(f"\n Finish time: {datetime.now()}")
print("=" * 60)
```
